```python
import pandas as pd
import sqlite3
import csv
```

```python
conexao_jogo_rapido = sqlite3.connect('jogo_rapido/data/jogo_rapido.db')
conexao_reportagens = sqlite3.connect('reportagens/data/reportagens.db')
```

```python
dados_jogo_rapido = pd.read_sql(con=conexao_jogo_rapido, sql='select * from artigos')
dados_reportagens  = pd.read_sql(con=conexao_reportagens, sql='select * from artigos')
```

```python
dados = pd.concat([dados_jogo_rapido, dados_reportagens])
```

```python
dados.to_csv('artigos_nucleo.csv', sep=';', encoding='utf-8-sig', index=False, quoting=csv.QUOTE_ALL)
```
